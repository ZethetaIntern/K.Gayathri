```python
import numpy as np
import pandas as pd

# Monte Carlo settings
np.random.seed(42)

n_scenarios = 1000
volatility = 0.01   # 100 bps = 1%

# Generate random parallel yield shifts
yield_shifts = np.random.normal(
    loc=0,
    scale=volatility,
    size=n_scenarios
)

# Check the generated scenarios
print("Number of Scenarios:", len(yield_shifts))
print("Minimum Yield Shift:", yield_shifts.min())
print("Maximum Yield Shift:", yield_shifts.max())
print("Mean Yield Shift:", yield_shifts.mean())
```

```python
# Monte Carlo P&L using Duration + Convexity

# Use the portfolio values already calculated in the notebook
portfolio_mod_duration = 0.136974
portfolio_convexity = 0.342076

# Portfolio market value
portfolio_market_value = 1000000

# Duration + Convexity P&L
mc_pnl = (
    -portfolio_mod_duration * yield_shifts * portfolio_market_value
    + 0.5 * portfolio_convexity * (yield_shifts ** 2) * portfolio_market_value
)

# Store results
mc_results = pd.DataFrame({
    "Yield_Shift": yield_shifts,
    "PnL": mc_pnl
})

print("Monte Carlo P&L calculation completed!")
print()
print(mc_results.head())
print()
print("Maximum Gain:", mc_results["PnL"].max())
print("Maximum Loss:", mc_results["PnL"].min())
```

```python
# VaR and CVaR Calculation

# VaR at 95% and 99% confidence levels
var_95 = -np.percentile(mc_results["PnL"], 5)
var_99 = -np.percentile(mc_results["PnL"], 1)

# CVaR (Expected Shortfall)
cvar_95 = -mc_results.loc[
    mc_results["PnL"] <= np.percentile(mc_results["PnL"], 5),
    "PnL"
].mean()

cvar_99 = -mc_results.loc[
    mc_results["PnL"] <= np.percentile(mc_results["PnL"], 1),
    "PnL"
].mean()

# Print results
print("===== MONTE CARLO RISK SUMMARY =====")
print(f"VaR 95%  : ₹{var_95:,.2f}")
print(f"VaR 99%  : ₹{var_99:,.2f}")
print(f"CVaR 95% : ₹{cvar_95:,.2f}")
print(f"CVaR 99% : ₹{cvar_99:,.2f}")
print(f"Maximum Loss : ₹{-mc_results['PnL'].min():,.2f}")
print(f"Maximum Gain : ₹{mc_results['PnL'].max():,.2f}")
```

```python
import matplotlib.pyplot as plt

# P&L Distribution with VaR lines

plt.figure(figsize=(10, 6))

plt.hist(mc_results["PnL"], bins=40)

plt.axvline(
    -var_95,
    linestyle="--",
    linewidth=2,
    label="VaR 95%"
)

plt.axvline(
    -var_99,
    linestyle="--",
    linewidth=2,
    label="VaR 99%"
)

plt.title("Monte Carlo Portfolio P&L Distribution")
plt.xlabel("Portfolio P&L (₹)")
plt.ylabel("Frequency")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
# P&L vs Yield Change

plt.figure(figsize=(10, 6))

plt.scatter(
    mc_results["Yield_Shift"] * 10000,
    mc_results["PnL"],
    alpha=0.6
)

plt.axhline(0, linestyle="--", linewidth=1)

plt.title("Monte Carlo P&L vs Yield Change")
plt.xlabel("Yield Change (bps)")
plt.ylabel("Portfolio P&L (₹)")
plt.grid(alpha=0.3)

plt.show()
```

```python
# Duration-only vs Duration + Convexity P&L
# Extreme scenarios: yield shift > 150 bps

extreme = mc_results[
    mc_results["Yield_Shift"].abs() > 0.015
].copy()

# Duration-only P&L
extreme["Duration_Only_PnL"] = (
    -portfolio_mod_duration
    * extreme["Yield_Shift"]
    * portfolio_market_value
)

# Duration + Convexity P&L
extreme["Duration_Convexity_PnL"] = (
    -portfolio_mod_duration
    * extreme["Yield_Shift"]
    * portfolio_market_value
    + 0.5
    * portfolio_convexity
    * (extreme["Yield_Shift"] ** 2)
    * portfolio_market_value
)

print("===== EXTREME SCENARIO COMPARISON =====")
print("Number of extreme scenarios:", len(extreme))

print("\nAverage P&L:")
print("Duration Only:",
      round(extreme["Duration_Only_PnL"].mean(), 2))

print("Duration + Convexity:",
      round(extreme["Duration_Convexity_PnL"].mean(), 2))

print("\nMaximum Loss:")
print("Duration Only:",
      round(extreme["Duration_Only_PnL"].min(), 2))

print("Duration + Convexity:",
      round(extreme["Duration_Convexity_PnL"].min(), 2))

print("\nSample extreme scenarios:")
print(
    extreme[
        [
            "Yield_Shift",
            "Duration_Only_PnL",
            "Duration_Convexity_PnL"
        ]
    ].head(10)
)
```

# Monte Carlo Simulation – Non-Parallel Yield Curve Shifts

```python
# Day 6 - Three-Factor Monte Carlo Setup

import numpy as np
import pandas as pd

np.random.seed(42)

n_scenarios = 1000

# Factor volatilities
parallel_vol = 0.010       # 100 bps
twist_vol = 0.0006         # 6 bps
butterfly_vol = 0.0006     # 6 bps

# Correlation structure from project requirements
corr_matrix = np.array([
    [1.0, -0.3,  0.1],
    [-0.3, 1.0,  0.2],
    [0.1,  0.2,  1.0]
])

# Cholesky decomposition
L = np.linalg.cholesky(corr_matrix)

# Independent random factors
z = np.random.normal(
    size=(n_scenarios, 3)
)

# Correlated random factors
factors = z @ L.T

parallel_shocks = factors[:, 0] * parallel_vol
twist_shocks = factors[:, 1] * twist_vol
butterfly_shocks = factors[:, 2] * butterfly_vol

print("Three-factor Monte Carlo setup completed!")
print("Number of scenarios:", n_scenarios)
print("Parallel volatility:", parallel_vol)
print("Twist volatility:", twist_vol)
print("Butterfly volatility:", butterfly_vol)
```

```python
# Non-Parallel Yield Curve Shocks
# 10 Key Rate Buckets

tenors = np.array([0.25, 0.5, 1, 2, 3, 5, 7, 10, 20, 30])

# Normalised tenor position
x = tenors / tenors.max()

# Parallel + Twist + Butterfly shocks
yield_curve_shocks = np.zeros((n_scenarios, len(tenors)))

for i in range(n_scenarios):
    
    parallel = parallel_shocks[i]
    twist = twist_shocks[i]
    butterfly = butterfly_shocks[i]
    
    # Twist: short end moves opposite to long end
    twist_component = twist * (1 - 2 * x)
    
    # Butterfly: middle maturities move differently
    butterfly_component = butterfly * (4 * x * (1 - x))
    
    yield_curve_shocks[i, :] = (
        parallel
        + twist_component
        + butterfly_component
    )

# Convert to basis points for easier interpretation
yield_curve_shocks_bps = yield_curve_shocks * 10000

print("Yield curve shock matrix created!")
print("Shape:", yield_curve_shocks_bps.shape)
print("Tenor buckets:", tenors)
print("\nFirst scenario (bps):")
print(np.round(yield_curve_shocks_bps[0], 2))
```

```python
# 3-Factor Monte Carlo P&L using KRD

# KRD values from the earlier analysis
krd_values = np.array([
    0.012,
    0.018,
    0.035,
    0.072,
    0.091,
    0.125,
    0.110,
    0.095,
    0.055,
    0.030
])

# Portfolio market value used in the earlier Monte Carlo section
portfolio_mv = portfolio_market_value

# KRD-based P&L for every scenario
three_factor_pnl = -np.sum(
    krd_values * yield_curve_shocks * portfolio_mv,
    axis=1
)

# Store results
three_factor_results = pd.DataFrame({
    "Parallel_Shift": parallel_shocks,
    "Twist_Shift": twist_shocks,
    "Butterfly_Shift": butterfly_shocks,
    "PnL": three_factor_pnl
})

print("===== THREE-FACTOR MONTE CARLO =====")
print("Scenarios:", len(three_factor_results))
print("Maximum Gain: ₹{:,.2f}".format(
    three_factor_results["PnL"].max()
))
print("Maximum Loss: ₹{:,.2f}".format(
    three_factor_results["PnL"].min()
))

print("\nFirst 5 scenarios:")
print(three_factor_results.head())
```

```python
# VaR Comparison: Parallel-only vs Three-factor model

parallel_var_95 = -np.percentile(mc_results["PnL"], 5)
parallel_var_99 = -np.percentile(mc_results["PnL"], 1)

three_factor_var_95 = -np.percentile(
    three_factor_results["PnL"], 5
)
three_factor_var_99 = -np.percentile(
    three_factor_results["PnL"], 1
)

var_comparison = pd.DataFrame({
    "Model": [
        "Parallel-only",
        "Three-factor"
    ],
    "VaR_95": [
        parallel_var_95,
        three_factor_var_95
    ],
    "VaR_99": [
        parallel_var_99,
        three_factor_var_99
    ]
})

print("===== VaR COMPARISON =====")
print(var_comparison)

print("\nVaR 95% Difference:")
print(
    three_factor_var_95 - parallel_var_95
)

print("\nVaR 99% Difference:")
print(
    three_factor_var_99 - parallel_var_99
)
```

```python
# Twist Volatility Sensitivity Analysis

twist_volatilities = np.array([0.0003, 0.0004, 0.0005, 0.0006, 0.0007, 0.0008, 0.0010])

sensitivity_results = []

for tv in twist_volatilities:

    # Generate twist shocks with the same correlated factor
    twist_test = factors[:, 1] * tv

    # Build yield curve shocks
    test_curve_shocks = np.zeros((n_scenarios, len(tenors)))

    for i in range(n_scenarios):

        parallel = parallel_shocks[i]
        butterfly = butterfly_shocks[i]

        twist_component = twist_test[i] * (1 - 2 * x)
        butterfly_component = butterfly * (4 * x * (1 - x))

        test_curve_shocks[i, :] = (
            parallel
            + twist_component
            + butterfly_component
        )

    # Calculate P&L
    test_pnl = -np.sum(
        krd_values * test_curve_shocks * portfolio_mv,
        axis=1
    )

    # Calculate VaR
    test_var_95 = -np.percentile(test_pnl, 5)

    sensitivity_results.append({
        "Twist_Volatility_bps": tv * 10000,
        "VaR_95": test_var_95
    })

twist_sensitivity = pd.DataFrame(sensitivity_results)

print("===== TWIST VOLATILITY SENSITIVITY =====")
print(twist_sensitivity)
```

```python
# STEP 1 — Automatically find and load the project dataset

import os
import pandas as pd

print("Searching for project data files...\n")

data_files = []

for root, dirs, files in os.walk("."):
    for file in files:
        if file.lower().endswith((".csv", ".xlsx", ".xls")):
            data_files.append(os.path.join(root, file))

print("Files found:")
for i, file in enumerate(data_files):
    print(i, "->", file)

if len(data_files) == 0:
    print("\nNO CSV/EXCEL FILE FOUND.")
else:
    # Prefer files whose names look related to the project
    preferred = [
        f for f in data_files
        if any(word in os.path.basename(f).lower()
               for word in ["bond", "risk", "krd", "dataset", "data"])
    ]

    if preferred:
        selected_file = preferred[0]
    else:
        selected_file = data_files[0]

    print("\nSelected file:")
    print(selected_file)

    if selected_file.lower().endswith(".csv"):
        df = pd.read_csv(selected_file)
    else:
        df = pd.read_excel(selected_file)

    print("\nDataset loaded successfully!")
    print("Shape:", df.shape)
    print("\nColumns:")
    print(list(df.columns))

    display(df.head())
```

```python
# STEP 2 — Check dataset structure

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
for i, col in enumerate(df.columns):
    print(i, "->", col)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# STEP 3 — Identify numerical variables

numeric_cols = df.select_dtypes(include="number").columns.tolist()

print("Numerical columns:")
for col in numeric_cols:
    print("-", col)

print("\nNumber of numerical columns:", len(numeric_cols))
```

```python
# STEP 4 — Identify categorical variables

categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Categorical columns:")
for col in categorical_cols:
    print("-", col)

print("\nNumber of categorical columns:", len(categorical_cols))
```

```python
# STEP 5 — Find columns that may represent risk/target

possible_targets = []

for col in df.columns:
    name = str(col).lower()

    if any(word in name for word in [
        "risk", "category", "class", "rating", "target", "label"
    ]):
        possible_targets.append(col)

print("Possible target columns:")
print(possible_targets)

for col in possible_targets:
    print("\n---", col, "---")
    print(df[col].value_counts(dropna=False).head(20))
```

```python
# STEP 6 — Clean duplicate rows and missing values

df = df.drop_duplicates().copy()

for col in numeric_cols:
    df[col] = df[col].replace(
        [float("inf"), float("-inf")],
        pd.NA
    )
    df[col] = df[col].fillna(df[col].median())

print("Cleaning completed.")
print("Final shape:", df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum().sum())
```

```python
# STEP 7 — Descriptive statistics

display(
    df[numeric_cols].describe().T
)
```

```python
# STEP 8 — Correlation matrix

correlation = df[numeric_cols].corr()

display(correlation.round(3))
```

```python
# STEP 9 — Prepare numerical feature matrix

X_numeric = df[numeric_cols].copy()

print("X_numeric shape:", X_numeric.shape)
print("Features:")
print(list(X_numeric.columns))
```

```python
# STEP 10 — Save cleaned dataset

cleaned_file = "Bond_Risk_KRD_Cleaned.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved as:")
print(cleaned_file)
```

```python
# STEP 11 — Select target column automatically

possible_targets = [
    col for col in df.columns
    if any(word in str(col).lower()
           for word in ["risk", "category", "class", "rating", "target", "label"])
]

print("Possible target columns:")
print(possible_targets)

if len(possible_targets) == 0:
    print("\nNo obvious target column found.")
else:
    target_col = possible_targets[0]
    print("\nSelected target:", target_col)
    print("\nTarget values:")
    print(df[target_col].value_counts(dropna=False))
```

```python
# STEP 12 — Prepare X and y

if len(possible_targets) == 0:
    print("STOP: No target column found.")
else:
    target_col = possible_targets[0]

    feature_cols = [
        col for col in numeric_cols
        if col != target_col
    ]

    X = df[feature_cols].copy()
    y = df[target_col].copy()

    print("Target:", target_col)
    print("Features:", feature_cols)
    print("X shape:", X.shape)
    print("y shape:", y.shape)
```

```python
# STEP 13 — Split data

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# STEP 14 — Feature scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
print("Training features:", X_train_scaled.shape)
print("Testing features:", X_test_scaled.shape)
```

```python
# STEP 15 — Prepare target for classification

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y.astype(str))

print("Target classes:")
for i, cls in enumerate(label_encoder.classes_):
    print(i, "->", cls)

print("\nNumber of classes:", len(label_encoder.classes_))
```

```python
# STEP 16 — Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model.fit(X_train_scaled, y_encoded[:len(X_train)])

# Use the same train/test split indices safely
_, _, y_train_enc, y_test_enc = train_test_split(
    X, y_encoded,
    test_size=0.20,
    random_state=42
)

log_model.fit(X_train_scaled, y_train_enc)

log_pred = log_model.predict(X_test_scaled)

log_accuracy = accuracy_score(y_test_enc, log_pred)

print("Logistic Regression Accuracy:",
      round(log_accuracy * 100, 2), "%")
```

```python
# STEP 17 — Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train_enc)

rf_pred = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test_enc, rf_pred)

print("Random Forest Accuracy:",
      round(rf_accuracy * 100, 2), "%")
```

```python
# STEP 18 — Corrected Model Evaluation

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

# Make sure predictions are available
if rf_accuracy >= log_accuracy:
    best_model = "Random Forest"
    best_pred = rf_pred
else:
    best_model = "Logistic Regression"
    best_pred = log_pred

# All possible encoded labels
all_labels = list(range(len(label_encoder.classes_)))

print("=" * 50)
print("MODEL COMPARISON")
print("=" * 50)

print("Logistic Regression Accuracy:",
      round(log_accuracy * 100, 2), "%")

print("Random Forest Accuracy:",
      round(rf_accuracy * 100, 2), "%")

print("\nBest Model:", best_model)

print("\n" + "=" * 50)
print("CLASSIFICATION REPORT")
print("=" * 50)

print(
    classification_report(
        y_test_enc,
        best_pred,
        labels=all_labels,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)

print("\n" + "=" * 50)
print("CONFUSION MATRIX")
print("=" * 50)

cm = confusion_matrix(
    y_test_enc,
    best_pred,
    labels=all_labels
)

print(cm)
```

```python
# STEP 19 — Confusion Matrix Visualization

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

plt.figure(figsize=(7, 5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

disp.plot()
plt.title("Confusion Matrix - " + best_model)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# STEP 20 — Random Forest Feature Importance

importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("Feature Importance:")
display(importance_df)

plt.figure(figsize=(9, 5))
plt.bar(
    importance_df["Feature"],
    importance_df["Importance"]
)
plt.title("Random Forest Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# STEP 21 — Actual vs Predicted

actual_risk = label_encoder.inverse_transform(y_test_enc)
predicted_risk = label_encoder.inverse_transform(best_pred)

results = pd.DataFrame({
    "Actual_Risk": actual_risk,
    "Predicted_Risk": predicted_risk
})

print("Actual vs Predicted Risk:")
display(results.head(20))
```
