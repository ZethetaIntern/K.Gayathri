## Step 12: Correlation Analysis

Measure relationships among important numerical variables.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.imshow(corr, cmap="coolwarm", interpolation="nearest")

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)

plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Heatmap")

plt.show()
```

```python
def classify_risk(ytm):
    if ytm < 0.05:
        return "Low Risk"
    elif ytm < 0.08:
        return "Medium Risk"
    else:
        return "High Risk"

bond_data["RiskCategory"] = bond_data["YieldToMaturity"].apply(classify_risk)
```

# Bond Risk Analysis Project

## Objective
The objective of this project is to analyze bond portfolio data and classify bonds into different risk categories based on Yield to Maturity.

## Dataset
- Total Records: 300
- Total Features: 44

## Data Cleaning
- Checked missing values
- CallDate contains missing values because many bonds are not callable.
- No duplicate records found.

## Exploratory Data Analysis
Performed:
- Distribution Analysis
- Box Plots
- Correlation Analysis
- Risk Category Distribution

## Risk Classification
Risk categories were created based on Yield to Maturity:
- Low Risk
- Medium Risk
- High Risk

## Key Insights
- Medium Risk bonds form the largest portion of the portfolio.
- High Risk bonds provide higher average yields.
- Modified Duration and Convexity show strong positive correlation.
- Market Value differs across risk categories.

## Conclusion
This project helps investors understand bond portfolio risk and supports better investment decisions.

```python
import matplotlib.pyplot as plt

corr = bond_data[[
    "YieldToMaturity",
    "ModifiedDuration",
    "Convexity",
    "MarketValue_INR",
    "PortfolioWeight"
]].corr()

plt.figure(figsize=(8,6))

plt.imshow(corr, cmap="coolwarm", interpolation="nearest")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Heatmap")

plt.show()
```

## Step 13: Correlation Insights

Interpret the correlation results obtained from the heatmap.

## Correlation Insights

- Modified Duration and Convexity have a strong positive correlation.
- Yield to Maturity has a negative correlation with Modified Duration.
- Market Value and Portfolio Weight are positively correlated.
- Correlation analysis helps identify relationships between bond characteristics.

## Step 14: Risk Categorization

Classify bonds into Low, Medium, and High Risk categories based on Yield to Maturity.

```python
bond_data.groupby("RiskCategory").agg({
    "YieldToMaturity":"mean",
    "ModifiedDuration":"mean",
    "MarketValue_INR":"mean",
    "PortfolioWeight":"sum"
}).round(2)
```

## Step 15: Risk Category Distribution

Visualize the percentage of Low, Medium, and High Risk bonds.

```python
import matplotlib.pyplot as plt

bond_data["RiskCategory"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)

plt.title("Risk Category Distribution")
plt.ylabel("")
plt.show()
```

## Step 16: Average Market Value by Risk Category

Compare the average market value across different risk categories.

```python
bond_data.groupby("RiskCategory")["MarketValue_INR"].mean().plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Average Market Value by Risk Category")
plt.ylabel("Market Value")
plt.show()
```

## Step 17: Risk Summary Table

Summarize key metrics for each risk category.

```python
risk_summary = bond_data.groupby("RiskCategory").agg({
    "YieldToMaturity":"mean",
    "ModifiedDuration":"mean",
    "MarketValue_INR":"mean",
    "PortfolioWeight":"sum"
}).round(2)

risk_summary
```
